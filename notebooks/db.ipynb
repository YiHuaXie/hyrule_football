```python
from sqlalchemy import inspect
from hyrule_football.database import engine, init_db
# 导入后模型会注册到 Base
from hyrule_football.models import Company, Match 

init_db()

def list_all_tables() -> str:
    inspector = inspect(engine)
    tables = inspector.get_table_names()

    if not tables:
        return "数据库中没有表"

    result = "数据库中的表:\n" + "\n".join(f"  - {table}" for table in tables)
    return result

result = list_all_tables()
print(result)  


```

```python
from hyrule_football.config import settings
from hyrule_football.schema.company import Company
import json 

def import_company_data():
    file_path = "/Users/neroxie/Desktop/博彩公司列表"
    output_path = "../hyrule_football/data/company.json"
    with open(file_path, "r", encoding="utf-8") as f:
        data = json.load(f)["data"]["companyList"]
    
    company_list = [Company(**x) for row in data.values() for x in row]
    unique = {c.cid: c.model_dump() for c in company_list}.values()

    with open(output_path, "w", encoding="utf-8") as f:
        json.dump(list(unique), f, ensure_ascii=False, indent=2)


import_company_data()
```
