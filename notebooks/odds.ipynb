```python
from hyrule_football.standard_odds import odds_serialization_from_markdown, asia_filter_odds

standard_odds_list = odds_serialization_from_markdown("../docs/odds/yj95_odds.md")
```

```python
from typing import List
from hyrule_football.standard_odds import StandardOdds

def fliter_standard_odds(
    odds_list: List[StandardOdds],
    system: str,
    goal_line: str,
    water_level: str,
) -> List[StandardOdds]:
    print(f"体系: {system}, 让球: {goal_line}, 水位: {water_level}")
    return asia_filter_odds(
        odds_list, system=system, goal_line=goal_line, water_level=water_level
    )
```

```python
from hyrule_football.standard_odds import HADOdds, HADOddsRange
import json

standard_odds = fliter_standard_odds(standard_odds_list, "意甲95", "-1.25", "低")

had_odds = HADOdds.odds_list(standard_odds)
for odds in had_odds:
    print(json.dumps(odds.model_dump(), indent=2, ensure_ascii=False))

had_odds_range = HADOddsRange.from_odds_list(had_odds)
print(json.dumps(had_odds_range.model_dump(), indent=2, ensure_ascii=False))

standard_odds = fliter_standard_odds(standard_odds_list, "意甲95", "-0.5", "低")
```
