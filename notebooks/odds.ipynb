```python
from hyrule_football.core.odds_engine import OddsEngine

standard_odds_list = OddsEngine.hyr_odds_serialization("../docs/odds/yj95_odds.md")
```

```python
from hyrule_football.third.ouhe.ouhe_odds import OHAsiaOdds, OHEuroOdds
from hyrule_football.core.odds_pattern import generate_euro_odds_pattern
from hyrule_football.core.odds_engine import OddsEngine

oh_euro_odds = OHEuroOdds(
    cid="26",
    matchId="123456",
    initOddsWin= "1.53",
    initOddsDraw= "3.90",
    initOddsLose= "6.00",
    initReturnRates= "92.88",
    nowOddsWin= "1.55",
    nowOddsDraw= "3.80",
    nowOddsLose= "5.80",
    nowReturnRates= "92.53",
)

oh_asia_odds = OHAsiaOdds(
    cid="26",
    matchId="123456",
    initBet="一球",
    initOddsWin="1.91",
    initOddsLose="1.83",
    initReturnRates="93.46", 
    nowBet="一球", 
    nowOddsWin="1.96", 
    nowOddsLose="1.81", 
    nowReturnRates="94.10",
)

print(f"基于主队视角, 盘口：{oh_asia_odds.initBet}, 赔率: {oh_asia_odds.initOddsWin}")
print(f"标准化术语：让球盘口: {oh_asia_odds.initGoalLine}, 水位: {oh_asia_odds.initWaterLevel}")

init_standard_odds, now_standard_odds = OddsEngine.hyr_standard_odds_from_oh_asia(standard_odds_list, oh_asia_odds, "意甲95")

init_euro_range = OddsEngine.hyr_euro_odds_range(init_standard_odds)
now_euro_range = OddsEngine.hyr_euro_odds_range(now_standard_odds)

print("初始欧指标准范围")
print(init_euro_range.model_dump())
print("即时欧指标准范围")
print(now_euro_range.model_dump())

init_euro_odds, now_euro_odds = OddsEngine.hyr_euro_odds_from_oh(oh_euro_odds)

init_pattern = generate_euro_odds_pattern(init_euro_odds, init_euro_range)
now_pattern = generate_euro_odds_pattern(now_euro_odds, now_euro_range)

print("初始格局")
print(init_pattern)
print("即时格局")
print(now_pattern)  
```
