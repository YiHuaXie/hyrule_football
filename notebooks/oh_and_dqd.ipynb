```python
import hyrule_football.service.dqd_service as dqd
from hyrule_football.schemas import DQDMatch
from datetime import datetime, timedelta

import json

yesterday_start = (datetime.now() - timedelta(days=1)).replace(hour=0, minute=0, second=0, microsecond=0)
today_start = datetime.now().replace(hour=0, minute=0, second=0, microsecond=0)
yesterday_start_str = yesterday_start.strftime("%Y-%m-%d %H:%M:%S")
today_start_str = yesterday_start.strftime("%Y-%m-%d %H:%M:%S")


date_start_list = [yesterday_start, today_start]
dqd_match_list = []
for date_start in date_start_list:
    date_start_str = date_start.strftime("%Y-%m-%d %H:%M:%S")
    tmp_match_list = dqd.request_played_match_list(date_start_str)
    dqd_match_list.extend(tmp_match_list)

print(f"从懂球帝接口中获取到共 {len(dqd_match_list)} 场比赛")
dqd_match_list = [DQDMatch(**m) for m in dqd_match_list]
dqd_match_list = [dqd_m for dqd_m in dqd_match_list if dqd_m.competition_name in ["英超", "西甲", "德甲", "意甲", "法甲"]]
print(f"五大联赛共 {len(dqd_match_list)} 场比赛")

for m in dqd_match_list:
    print(json.dumps(m.model_dump(), indent=4, ensure_ascii=False))
```
